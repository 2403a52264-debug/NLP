# Corpus

```python
D1 = "I am a professinal coder"

D2 = "I have much experience in training the machine learning models"

D3 = "I am studing in sr university"

D4 = "I was very intersted in playing games"

D5 = "I also very intersted in creating content"

D6 = "I was also very interseted in kite flying competitions"

D7 =  "I drink water 4 literes per day"

D8 = "I got 32 marks in computer networks subject"

D9 = "I have B grade cgpa"

D10 =  "I was very excited about my life"
```

# We need to import the required libraries

```python
import nltk
import spacy
nltk.download('punkt_tab')
import collections
```

# Unigram counts

```python
combine_corpus = D1 + D2  + D3 + D5 + D6 + D7 + D8 + D9 + D10
combined_text = combine_corpus.lower()

tokens = nltk.word_tokenize(combined_text)

from nltk.probability import FreqDist

fdist_unigrams = FreqDist(tokens)

print("Unigram Counts:")
for unigram, count in fdist_unigrams.most_common():
    print(f"{unigram}: {count}")
```

# Bi-gram counts

```python
bigrams = list(nltk.bigrams(tokens))
fdist_bigrams = FreqDist(bigrams)

print("\nBigram Counts:")
for bigram, count in fdist_bigrams.most_common():
    print(f"{bigram}: {count}")
```

# Trigram counts

```python
trigrams = list(nltk.trigrams(tokens))
fdist_trigrams = FreqDist(trigrams)

print("\nTrigram Counts:")
for trigram, count in fdist_trigrams.most_common():
    print(f"{trigram}: {count}")
```

# Next word prediction using bigram model

```python
bigram_probabilities = {}
for bigram, count in fdist_bigrams.items():
    w1, w2 = bigram
    if fdist_unigrams[w1] > 0:
        bigram_probabilities[bigram] = count / fdist_unigrams[w1]
    else:
        bigram_probabilities[bigram] = 0.0

print("\nBigram Probabilities:")
for bigram, prob in list(bigram_probabilities.items())[:10]:
    print(f"{bigram}: {prob:.4f}")
```

# Deployment of bigram model

```python
def predict_next_word_bigram(start_word, bigram_probabilities):
    start_word = start_word.lower()
    possible_next_words = {}
    for (w1, w2), prob in bigram_probabilities.items():
        if w1 == start_word:
            possible_next_words[w2] = prob

    if not possible_next_words:
        return "No prediction available for this word based on bigram model."

    best_next_word = max(possible_next_words, key=possible_next_words.get)
    return best_next_word

ip_text=input("enter text")
next_word1 = predict_next_word_bigram(ip_text, bigram_probabilities)
print(f"Given sequence: '{ip_text}', predicted next word: '{next_word1}'")
```

# Next word prediction using trigram model

```python
trigram_probabilities = {}
for trigram, count in fdist_trigrams.items():
    w1, w2, w3 = trigram
    if (w1, w2) in fdist_bigrams and fdist_bigrams[(w1, w2)] > 0:
        trigram_probabilities[trigram] = count / fdist_bigrams[(w1, w2)]
    else:
        trigram_probabilities[trigram] = 0.0

print("\nTrigram Probabilities (sample):")
for trigram, prob in list(trigram_probabilities.items())[:10]:
    print(f"{trigram}: {prob:.4f}")

def predict_next_word_trigram(word1, word2, trigram_probabilities):
    word1 = word1.lower()
    word2 = word2.lower()
    possible_next_words = {}
    for (w1_t, w2_t, w3), prob in trigram_probabilities.items():
        if w1_t == word1 and w2_t == word2:
            possible_next_words[w3] = prob

    if not possible_next_words:
        return "No prediction available for this sequence based on trigram model."

    best_next_word = max(possible_next_words, key=possible_next_words.get)
    return best_next_word
```

# Deployment of trigram model

```python
ip_text1 = input("Enter first word: ")
ip_text2 = input("Enter second word: ")

next_word_trigram = predict_next_word_trigram(ip_text1, ip_text2, trigram_probabilities)
print(f"Given sequence: '{ip_text1} {ip_text2}', predicted next word: '{next_word_trigram}'")
```

# Next Word Prediction Using Bi-Gram Counts with Laplace Smoothening

```python
V = len(fdist_unigrams)

smoothed_bigram_probabilities = {}

for bigram in bigrams:
    w1, w2 = bigram
    numerator = fdist_bigrams.get(bigram, 0) + 1
    denominator = fdist_unigrams.get(w1, 0) + V

    smoothed_bigram_probabilities[bigram] = numerator / denominator

print("\nSmoothed Bigram Probabilities (sample):")
for bigram, prob in list(smoothed_bigram_probabilities.items())[:10]:
    print(f"{bigram}: {prob:.4f}")

def predict_next_word_bigram_laplace(start_word, smoothed_probs):
    start_word = start_word.lower()
    possible_next_words = {}
    for (w1, w2), prob in smoothed_probs.items():
        if w1 == start_word:
            possible_next_words[w2] = prob

    if not possible_next_words:
        return "No prediction available for this word based on smoothed bigram model."

    best_next_word = max(possible_next_words, key=possible_next_words.get)
    return best_next_word
```

# Deployment of Laplace Smoothening based Bi-Gram Model

```python
ip_text_smoothed = input("Enter a *single* word for Laplace smoothed bigram prediction: ")
next_word_smoothed = predict_next_word_bigram_laplace(ip_text_smoothed, smoothed_bigram_probabilities)
print(f"Given sequence: '{ip_text_smoothed}', predicted next word (Laplace smoothed): '{next_word_smoothed}'")
```

## Next Word Prediction Using Tri-Gram Counts with Laplace Smoothening

```python
smoothed_trigram_probabilities = {}

for trigram in trigrams:
    w1, w2, w3 = trigram
    bigram_count = fdist_bigrams.get((w1, w2), 0)

    numerator = fdist_trigrams.get(trigram, 0) + 1
    denominator = bigram_count + V

    smoothed_trigram_probabilities[trigram] = numerator / denominator

print("\nSmoothed Trigram Probabilities (sample):")
for trigram, prob in list(smoothed_trigram_probabilities.items())[:10]:
    print(f"{trigram}: {prob:.4f}")

def predict_next_word_trigram_laplace(word1, word2, smoothed_probs):
    word1 = word1.lower()
    word2 = word2.lower()
    possible_next_words = {}
    for (w1_t, w2_t, w3), prob in smoothed_probs.items():
        if w1_t == word1 and w2_t == word2:
            possible_next_words[w3] = prob

    if not possible_next_words:
        return "No prediction available for this sequence based on smoothed trigram model."

    best_next_word = max(possible_next_words, key=possible_next_words.get)
    return best_next_word
```

## Deployment of Laplace Smoothening based Tri-Gram Model

```python
ip_text1_smoothed = input("Enter *first single word* for Laplace smoothed trigram prediction: ")
ip_text2_smoothed = input("Enter *second single word* for Laplace smoothed trigram prediction: ")

next_word_trigram_smoothed = predict_next_word_trigram_laplace(ip_text1_smoothed, ip_text2_smoothed, smoothed_trigram_probabilities)
print(f"Given sequence: '{ip_text1_smoothed} {ip_text2_smoothed}', predicted next word (Laplace smoothed): '{next_word_trigram_smoothed}'")
```

## Next Word Prediction Using Bi-Gram Counts with Add-K Smoothening

```python
k = 1
V = len(fdist_unigrams)

smoothed_bigram_probabilities_addk = {}

for bigram in bigrams:
    w1, w2 = bigram
    numerator = fdist_bigrams.get(bigram, 0) + k
    denominator = fdist_unigrams.get(w1, 0) + k * V

    smoothed_bigram_probabilities_addk[bigram] = numerator / denominator

print(f"\nSmoothed Bigram Probabilities (Add-{k} Smoothing, sample):")
for bigram, prob in list(smoothed_bigram_probabilities_addk.items())[:10]:
    print(f"{bigram}: {prob:.4f}")

def predict_next_word_bigram_addk(start_word, smoothed_probs_addk):
    start_word = start_word.lower()
    possible_next_words = {}
    for (w1, w2), prob in smoothed_probs_addk.items():
        if w1 == start_word:
            possible_next_words[w2] = prob

    if not possible_next_words:
        return "No prediction available for this word based on Add-K smoothed bigram model."

    best_next_word = max(possible_next_words, key=possible_next_words.get)
    return best_next_word
```

## Next Word Prediction Using Tri-Gram Counts with Add-K Smoothening

```python
k = 1
V = len(fdist_unigrams)

smoothed_trigram_probabilities_addk = {}

for trigram in trigrams:
    w1, w2, w3 = trigram
    bigram_count = fdist_bigrams.get((w1, w2), 0)

    numerator = fdist_trigrams.get(trigram, 0) + k
    denominator = bigram_count + k * V

    smoothed_trigram_probabilities_addk[trigram] = numerator / denominator

print(f"\nSmoothed Trigram Probabilities (Add-{k} Smoothing, sample):")
for trigram, prob in list(smoothed_trigram_probabilities_addk.items())[:10]:
    print(f"{trigram}: {prob:.4f}")

def predict_next_word_trigram_addk(word1, word2, smoothed_probs_addk):
    word1 = word1.lower()
    word2 = word2.lower()
    possible_next_words = {}
    for (w1_t, w2_t, w3), prob in smoothed_probs_addk.items():
        if w1_t == word1 and w2_t == word2:
            possible_next_words[w3] = prob

    if not possible_next_words:
        return "No prediction available for this sequence based on Add-K smoothed trigram model."

    best_next_word = max(possible_next_words, key=possible_next_words.get)
    return best_next_word
```
